# Make a Configuration File

Builds the YAML configuration file used by the automated **sixdegrees** BRS analysis notebooks. This notebook:

- Defines the event time window (`tbeg`, `tend`), origin time, magnitude, and project name
- Sets LIGO **LIV** station coordinates and SEED IDs for translation and rotation channels (plus any dummy traces)
- Specifies where to read MiniSEED data and station inventories, and how translation data are corrected (`tra_remove_response`, `tra_output`, `resample_rate`)
- Configures output paths, units, verbosity, and optional ZNE rotation behavior
- Instantiates `sixdegrees` with the dict and writes the config to `./config/<project>.yaml` via `store_as_yaml`

Edit the `config` dictionary for each event, then run the cells below to generate the file consumed by later pipeline steps.

```python
from sixdegrees.sixdegrees import sixdegrees
```

## Configurations

Adjust the configurations for the specific event

```python
config = {

    # Required time window parameters
    'tbeg': "2020-07-22 06:12:44",  # Start time
    'tend': "2020-07-22 08:12:44",  # End time

    # Event information
    'origin_time': "2020-07-22 06:12:44",
    'magnitude': 7.8,

    # Project information
    'project': "2020_07_22",  # Project name

    # Station coordinates (required)
    'station_lon': -90.77424,  # LIGO LIV longitude
    'station_lat': 30.56289,    # LIGO LIV latitude

    # SEED identifiers
    'seed': "XX.LIV..",     # Default seed ID
    'rot_seed': ["LI.LIV.IY.BJX", "LI.LIV.IX.BJY"], # Rotation sensor seed ID
    'tra_seed': ["LI.LIV.IX.BHZ", "LI.LIV.IX.BHX", "LI.LIV.IX.BHY"],  # Translation sensor seed ID

    # dummy traces
    'dummy_trace': ["LI.LIV.IX.BJZ"],
    
    # Data source
    'data_source': "mseed_file",  # "fdsn" | "sds" | "mseed_file"

    # Path to local mseed file
    'path_to_mseed_file': "./data/2020_07_22.mseed",  # "./data/eventM62_6c.mseed",

    # Path to inventory of rotation and translation sensor
    'path_to_inv_rot': "./metadata/LIGO_HAN_LIV_AB.xml",  # Rotation station inventory
    'path_to_inv_tra': "./metadata/LIGO_HAN_LIV_AB.xml",  # Translation station inventory

    # Remove response of translation data
    'tra_remove_response': True,
    'tra_output': "VEL",

    # Resample rate
    'resample_rate': 10,  # Hz

    # Path to figures
    'path_to_figures': "./figures/",

    # Path configurations
    'workdir': "./",            # Working directory
    'path_to_data_out': "./output/",  # Output directory

    # Set rotation of components
    'rotate_zne': False,

    # Rotational and translational units
    'runit': 'rad',          # Rotational units
    'tunit': 'm/s',          # Translational units

    # Optional parameters
    'verbose': True,            # Print detailed information
}
```

Create a instance of sixdegrees

```python
sd = sixdegrees(conf=config)
```

Use the internal function to store the configuration as a YAML file

```python
sd.store_as_yaml(config, f"./config/{config['project']}")
```
